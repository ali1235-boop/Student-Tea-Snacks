```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('student_tea_snacks.csv')

print(df.head())
```

```python
print("Total number of students:", len(df))

print("\nDataset information:")
print(df.info())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
tea_students = df[df['prefers_tea'] == 'Yes']

tea_counts = tea_students['tea_type'].value_counts()

print("Tea Preferences:")
print(tea_counts)

print("\nMost Preferred Tea:")
print(tea_counts.idxmax())
```

```python
gender_average = df.groupby('gender')['cups_per_day'].mean()

print("Average Cups Per Day by Gender:")
print(gender_average)
```

```python
snack_counts = df['snack_choice'].value_counts()

print("Snack Choices:")
print(snack_counts)

snack_counts.plot(kind='bar', figsize=(8, 5))

plt.title('Student Snack Preferences')
plt.xlabel('Snack Choices')
plt.ylabel('Number of Students')

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
tea_students = df[df['prefers_tea'] == 'Yes']

age_comparison = pd.crosstab(
    tea_students['age_group'],
    tea_students['tea_type']
)

print("Tea Preference by Age Group:")
print(age_comparison)
```

```python
age_comparison.plot(kind='bar', figsize=(8, 5))

plt.title('Tea Preferences by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Students')

plt.xticks(rotation=0)
plt.legend(title='Tea Type')
plt.tight_layout()
plt.show()
```

```python
non_tea_students = df[df['prefers_tea'] == 'No']

print("Number of students who don't drink tea:",
      len(non_tea_students))
```
